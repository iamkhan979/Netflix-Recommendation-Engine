```python
# movie count 
Movie_count= df.isna().sum()['Ratings']
Movie_count
```

```python
# total unique customer
total_uniq = df['cust_id'].nunique()
total_uniq
```

```python
cust_count = total_uniq-Movie_count
cust_count
```

```python
Ratings_count= df['Ratings'].count() - Movie_count
Ratings_count
```

```python
stars_count = df['Ratings'].value_counts()
stars_count
```

```python
# creating movie id column
movie_id = None
movie_list = []
for customer in df['cust_id']:
    if ":" in customer:
        movie_id = int(customer.replace(':',""))
    movie_list.append(movie_id)

                       
```

```python
movie_list
```

```python
df['Movie_id']= movie_list
df
```

```python
df.dropna(inplace=True)
```

```python
df
```

```python
df.info()
```

```python
df['cust_id'] = df['cust_id'].astype('int')
df.info()
```

```python
movie_rat_count=df['Movie_id'].value_counts()
movie_rat_count
```

```python
bench_marks1=round(movie_rat_count.quantile(0.6),0)
bench_marks1
```

```python
# what movies id we need to remove .
drop_movie_index=movie_rat_count[movie_rat_count<bench_marks1].index
drop_movie_index
```

```python
cust_rat_count=df['cust_id'].value_counts()
cust_rat_count
```

```python
bench_marks2=round(cust_rat_count.quantile(0.6),0)
bench_marks2
```

```python
drop_cust_index=cust_rat_count[cust_rat_count<bench_marks2].index
drop_cust_index
```

```python
drop_movie_index,drop_cust_index
```

```python
df=df[~df['Movie_id'].isin(drop_movie_index)]
df=df[~df['cust_id'].isin(drop_cust_index)]
```
